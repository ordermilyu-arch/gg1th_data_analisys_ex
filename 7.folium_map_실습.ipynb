```python
import folium
#folium import하여 사용
```

```python
# dir(folium)
# folium 라이브러리 안의 클래스들 확인
```

```python
# 디폴트 위경도 지정(포커싱), 지도사이트 Map 객체 생성
folium_map = folium.Map(location=[37.5662952, 126.9779451], zoom_start=16)

# folium.map = 지도 객체를 만드는 클래스
# folium.map(loacation[x,y]) = 지도 객체를 만들때 중심을 어디에 둘지 정하는 변수 , 이때 x는 위도, y는 경도
# folium.map(zoom_start=c) = c는 미리 정해진 zoom level 에 맞추어 진행 이때 숫자가 작아질수록 더 멀리서보는(축소), 16의 경우 동네/주변 지역
# 즉 위도 37.5662952, 경도 126.9779451 위치를 중심으로 지도 객체를 만들라는 내용
# dir(folium.map)
```

```python
folium.Marker(location=[37.5662952, 126.9779451]).add_to(folium_map)

# folium.Marker = 마커(핀)울 만드는 클래스
# add_to(d)는 앞에 생성한 마커를 d맵에 추가하라는 의미
```

```python
popup = folium.Popup('서울시청 위치', max_width=300)
folium.Marker(location=[37.5662952, 126.9779451], popup=popup).add_to(folium_map)

# popup은 정해진 마커를 눌렀을때 어떤 정보가 나오게 할지에 대한 내용
# max_width는 팝업의 px사이즈(가로)를 나타냄
```
